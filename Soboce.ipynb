```python
# ============================================================
# CELDA 2: Carga de datos (detecta .xls, .xlsx o .csv)
# ============================================================
from pathlib import Path
import pandas as pd

CARPETA_DATA = Path("../Data")

# Buscar el archivo automáticamente
archivos = (
    list(CARPETA_DATA.glob("*.xlsx")) +
    list(CARPETA_DATA.glob("*.xls")) +
    list(CARPETA_DATA.glob("*.csv"))
)

# Filtrar los que tengan "Diesel" en el nombre
archivos_diesel = [f for f in archivos if "diesel" in f.name.lower()]

if not archivos_diesel:
    raise FileNotFoundError(
        f"❌ No se encontró ningún archivo con 'Diesel' en {CARPETA_DATA.resolve()}\n"
        f"Archivos disponibles: {[f.name for f in archivos]}"
    )

ARCHIVO = archivos_diesel[0]
print(f"📂 Cargando: {ARCHIVO.name}")

# Cargar según extensión
if ARCHIVO.suffix == ".csv":
    df = pd.read_csv(ARCHIVO)
elif ARCHIVO.suffix == ".xlsx":
    df = pd.read_excel(ARCHIVO, engine="openpyxl")
elif ARCHIVO.suffix == ".xls":
    df = pd.read_excel(ARCHIVO, engine="xlrd")

# Procesar
df["Fecha"] = pd.to_datetime(df["Fecha"])
df = df.sort_values("Fecha").reset_index(drop=True)

df["Costo_Diesel_Bs_por_Ton"] = (
    df["Costo_Diesel_Total_Bs"] / df["Volumen_Produccion_Ton"]
)

COLS_KPI = [
    "Precio_Diesel_Aplicado_SOBOCE_Bs_Litro",
    "Volumen_Produccion_Ton",
    "Costo_Diesel_Total_Bs",
    "Costo_Diesel_Bs_por_Ton",
]

print(f"\n📊 Dimensiones: {df.shape}")
print(f"📅 Rango: {df['Fecha'].min().date()} → {df['Fecha'].max().date()}")
print(f"\n📋 Columnas:\n{df.columns.tolist()}")
df.head()
```
